# Tourism Experience Analytics - Exploratory Data Analysis & Insights

This notebook documents the exploratory data analysis (EDA), statistical distributions, attraction rating trends, demographic segmentation, and visit mode correlations.

```python
import os
import sys
sys.path.append('..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.data.preprocessor import build_consolidated_dataset
from src.features.build_features import build_all_features

%matplotlib inline
plt.style.use('ggplot')
```

## 1. Data Ingestion & Overview

```python
master_df, users_df, items_df = build_consolidated_dataset()
print(f"Master DataFrame shape: {master_df.shape}")
print(f"Users DataFrame shape: {users_df.shape}")
print(f"Items DataFrame shape: {items_df.shape}")
master_df.head()
```

## 2. Rating Distribution & Visit Mode Analysis

```python
plt.figure(figsize=(10, 4))
sns.countplot(data=master_df, x='Rating', palette='viridis')
plt.title('Distribution of User Ratings (1 to 5)')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(10, 4))
sns.countplot(data=master_df, x='VisitModeName', palette='mako')
plt.title('Distribution of Visitor Modes (Business, Family, Solo, Friends, Couples)')
plt.xlabel('Visit Mode')
plt.ylabel('Count')
plt.show()
```

## 3. Top Tourist Attractions & Categorical Distribution

```python
top_attractions = master_df.groupby('Attraction').agg(
    Avg_Rating=('Rating', 'mean'),
    Review_Count=('Rating', 'count')
).sort_values(by='Review_Count', ascending=False).head(10)

print('Top 10 Most Visited Attractions:')
display(top_attractions)
```
